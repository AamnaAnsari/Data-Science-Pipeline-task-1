```python
import sys
import os

sys.path.append(os.path.abspath('..'))

# custom class
from src.data_loader import DataLoader

# Pandas and plotting libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# DataLoader initializion
loader = DataLoader("../data/raw/dataset.csv")

# Data loading
df = loader.load_csv()

# display first five rows
df.head()
```

```python
# Data ki basic details aur types
print("--- Data Info ---")
df.info()

print("\n--- Missing Values Count ---")
# Har column mein kitni missing values hain, unka total
print(df.isnull().sum())
```

```python
# 1. Zaroori libraries import karein
import matplotlib.pyplot as plt
import seaborn as sns

# Jupyter notebook mein graph directly show karne ke liye magic command
%matplotlib inline

# 2. Graph ka code
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='TotalPurchases', hue='City', s=100)

plt.title('Age vs Total Purchases (City-wise)')
plt.xlabel('Age')
plt.ylabel('Total Purchases')
plt.show()
```

```python
# 1. Age column ki missing values ko uske 'Median' se fill karna
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)
print(f"Missing 'Age' ko {age_median} se fill kar diya gaya hai.")

# 2. LoyaltyScore ki missing values ko uske 'Mean' (average) se fill karna
loyalty_mean = df['LoyaltyScore'].mean()
# round(..., 2) use kiya hai taake score 2 decimal places tak rahe
df['LoyaltyScore'] = df['LoyaltyScore'].fillna(round(loyalty_mean, 2))
print(f"Missing 'LoyaltyScore' ko {round(loyalty_mean, 2)} se fill kar diya gaya hai.")

# 3. Check karte hain ke kya ab koi NaN bacha hai?
print("\n--- Data Cleaning Ke Baad Missing Values ---")
print(df.isnull().sum())

# Cleaned data ki shakal dekhein
df.head()
```

```python
# Age vs TotalPurchases ka scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='TotalPurchases', hue='City', s=100)

plt.title('Age vs Total Purchases (City-wise)')
plt.xlabel('Age')
plt.ylabel('Total Purchases')
plt.show()
```
