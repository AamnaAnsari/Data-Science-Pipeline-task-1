```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# For showing graphs in notebook
%matplotlib inline
```

```python
# Read Raw Data
df = pd.read_csv('../data/raw/customer_data.csv')
print(f"Data Shape: {df.shape}")
df.head()
```

```python
# Data mein koi null values ya ajeeb cheez toh nahi?
display(df.info())
display(df.describe())
```

```python
# Distributions of Age and TotalPurchases
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title('Customer Age Distribution')

plt.subplot(1, 2, 2)
sns.histplot(df['TotalPurchases'], bins=20, kde=True, color='salmon')
plt.title('Total Purchases Distribution')

plt.tight_layout()
plt.show()
```

```python
numeric_df = df[['Age', 'TotalPurchases', 'LoyaltyScore']]

plt.figure(figsize=(6, 4))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Feature Correlation Matrix')
plt.show()
```

```python
# Check if there are any missing values in our new dataset
print("Missing Values in Dataset:")
display(df.isnull().sum())
```

```python
#One-hot encoding step (Already done in feature Engineering)
df_cleaned = pd.get_dummies(df, columns=['City'], drop_first=True)
print("Data after Encoding Categorical Variables (Ready for Model):")
display(df_cleaned.head())
```
